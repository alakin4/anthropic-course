```python
# Load env variables
from dotenv import load_dotenv
from IPython.display import Markdown
load_dotenv()
```

```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# Make a request
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is context engineering? Answer in one sentence."
        }
    ]
)
```

```python
# display message.content[0].text as markdown
Markdown(message.content[0].text)
```
