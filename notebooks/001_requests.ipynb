```python
# Load env variables
from dotenv import load_dotenv
from IPython.display import Markdown
load_dotenv()
```

```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# Make a request
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is context engineering? Answer in one sentence."
        }
    ]
)
```

```python
# display message.content[0].text as markdown
Markdown(message.content[0].text)
```

### Multi-turn conversations

```python
# Helper functions to build messages
def add_user_message(messages, text):
    user_message = {
        "role": "user",
        "content": text
    }
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {
        "role": "assistant",
        "content": text
    }
    messages.append(assistant_message)

def chat(messages):
    response = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    return response.content[0].text
```

```python
# Make a starting list of messages

messages = []

# Add in the initial user question of "Define context engineering in one sentence."

add_user_message(messages, "Define context engineering in one sentence.")

# pass the list of messages into chat to get an answer

answer = chat(messages)

# Take the answer and add it as an assistant message into our list

add_assistant_message(messages, answer)

# Add in the user's follow-up question

add_user_message(messages, "What are its key principles")

# Call chat again with the list of messages to get a final answer
final_answer = chat(messages)
```
